```python
df = pd.read_parquet('green_tripdata_2025-11.parquet')
```

```python
start_date = '2025-11-01'
end_date = '2025-12-01'
filtered_df = df[(df['lpep_pickup_datetime'] >= start_date) & (df['lpep_pickup_datetime'] < end_date)]
filtered_df = filtered_df[filtered_df['trip_distance'] <= 1]
filtered_df
```

```python
longest_range = df[df['trip_distance'] < 100]
max_trip_distance = longest_range['trip_distance'].max()
row_with_max = longest_range[longest_range['trip_distance'] == max_trip_distance]
row_with_max
```

```python
zones = pd.read_csv('taxi_zone_lookup.csv')
```

```python
zones
```

```python
merged_df = df.merge(zones, left_on='PULocationID', right_on='LocationID', how='left')
```

```python
# Ensure datetime column is in datetime format
df['lpep_pickup_datetime'] = pd.to_datetime(df['lpep_pickup_datetime'])

# Filter for November 18th, 2025
filtered = df[df['lpep_pickup_datetime'].dt.date == pd.to_datetime('2025-11-18').date()]

# Group by PULocationID (or another column) and sum total_amount
grouped = filtered.groupby('PULocationID')['total_amount'].sum()

# Find the group with the largest sum
largest = grouped.idxmax()
largest_sum = grouped.max()

print(f"{zones[zones['LocationID'] == largest]['Zone'].values[0]} has the largest total_amount of {largest_sum} ")
```

```python
# Filter for East Harlem North and November 2025
merged_df['lpep_pickup_datetime'] = pd.to_datetime(merged_df['lpep_pickup_datetime'])
mask = (
    (merged_df['Zone'] == 'East Harlem North') &
    (merged_df['lpep_pickup_datetime'].dt.year == 2025) &
    (merged_df['lpep_pickup_datetime'].dt.month == 11)
)
filtered = merged_df[mask]

# Merge to get drop-off zone names
filtered = filtered.merge(zones, left_on='DOLocationID', right_on='LocationID', how='left', suffixes=('', '_dropoff'))

# Find the row with the largest tip_amount
max_tip_row = filtered.loc[filtered['tip_amount'].idxmax()]

# Get the drop-off zone
dropoff_zone = max_tip_row['Zone_dropoff']
print(f"Drop-off zone with largest tip: {dropoff_zone}")
```
